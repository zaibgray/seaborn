## Seaborn
### Types of Functions
- Figure Level
- Axis Level

## Seaborn Cheat Sheet for ML & Research

> **Syntax:** `sns.plot(data=df, x="x", y="y")`
>
> **Axes-level** (e.g. `scatterplot`): draws on one plot. Good for subplots.
>
> **Figure-level** (e.g. `relplot`): makes its own figure, handles the legend, and can split into multiple panels with `col=` / `row=`.

---

## 1. Relational Plots

| Plot                | Purpose                | Data                  |
| ------------------- | ----------------------- | --------------------- |
| `sns.lineplot()`    | Track trends / changes | Numerical + Numerical |
| `sns.scatterplot()` | Show relationships     | Numerical + Numerical |

```python
sns.lineplot(data=df, x="x", y="y")
sns.scatterplot(data=df, x="x", y="y")
```

### Figure-level: `sns.relplot()`

| Plot             | Purpose                                  | Data                   |
| ---------------- | ----------------------------------------- | ---------------------- |
| `sns.relplot()`  | Line/scatter plots with faceting support | Numerical + Numerical  |

```python
sns.relplot(data=df, x="x", y="y", kind="scatter", col="category")
sns.relplot(data=df, x="x", y="y", kind="line", row="category")
```

---

## 2. Distribution Plots

| Plot             | Purpose                     | Data      |
| ---------------- | ---------------------------- | --------- |
| `sns.histplot()` | Show distribution with bars | Numerical |
| `sns.kdeplot()`  | Show smooth distribution    | Numerical |

```python
sns.histplot(data=df, x="x")
sns.kdeplot(data=df, x="x")
```

### Figure-level: `sns.displot()`

| Plot             | Purpose                                        | Data      |
| ---------------- | ------------------------------------------------ | --------- |
| `sns.displot()`  | Hist/KDE/ECDF plots with faceting support        | Numerical |

```python
sns.displot(data=df, x="x", kind="hist", col="category")
sns.displot(data=df, x="x", kind="kde", hue="category")
sns.displot(data=df, x="x", kind="ecdf")
```

---

## 3. Categorical Plots

| Plot              | Purpose                    | Data                    |
| ----------------- | ---------------------------- | ------------------------ |
| `sns.barplot()`   | Compare values by category | Categorical + Numerical |
| `sns.boxplot()`   | Show spread & outliers     | Categorical + Numerical |
| `sns.countplot()` | Count each category        | Categorical             |

```python
sns.barplot(data=df, x="category", y="value")
sns.boxplot(data=df, x="category", y="value")
sns.countplot(data=df, x="category")
```

### Figure-level: `sns.catplot()`

| Plot             | Purpose                                                        | Data                    |
| ---------------- | ----------------------------------------------------------------- | ------------------------ |
| `sns.catplot()`  | Bar/box/violin/strip/swarm/count plots with faceting support     | Categorical + Numerical |

```python
sns.catplot(data=df, x="category", y="value", kind="bar", col="group")
sns.catplot(data=df, x="category", y="value", kind="box", hue="group")
sns.catplot(data=df, x="category", kind="count")
```

---

## 4. Regression Plots

| Plot              | Purpose                 | Data                  |
| ----------------- | ------------------------- | ---------------------- |
| `sns.regplot()`   | Show regression trend   | Numerical + Numerical |
| `sns.residplot()` | Show residuals / errors | Numerical + Numerical |

```python
sns.regplot(data=df, x="x", y="y")
sns.residplot(data=df, x="x", y="y")
```

### Figure-level: `sns.lmplot()`

| Plot            | Purpose                                        | Data                   |
| --------------- | ------------------------------------------------- | ----------------------- |
| `sns.lmplot()`  | Regression trend lines with faceting support      | Numerical + Numerical  |

```python
sns.lmplot(data=df, x="x", y="y", col="category")
sns.lmplot(data=df, x="x", y="y", hue="category")
```

---

## 5. Matrix Plots

| Plot            | Purpose                      | Data             |
| --------------- | ------------------------------ | ----------------- |
| `sns.heatmap()` | Show correlations / matrices | Numerical Matrix |

```python
sns.heatmap(data, annot=True)
```

> Note: `heatmap()` and `clustermap()` are both figure-managing; `clustermap()` is the closer figure-level equivalent (adds dendrograms + clustering).

```python
sns.clustermap(data, annot=True, cmap="viridis")
```

---

## 6. Multi-Plot Grids (already figure-level)

| Plot              | Purpose               | Data                    |
| ----------------- | ------------------------ | ------------------------ |
| `sns.pairplot()`  | Compare many features | Numerical + Categorical |
| `sns.jointplot()` | Explore two variables | Numerical + Numerical   |

```python
sns.pairplot(df)
sns.jointplot(data=df, x="x", y="y")
```

## Quick Data-Type Guide

| Data                        | Meaning                         |
| ---------------------------- | ---------------------------------- |
| **Numerical**               | Age, salary, score, price       |
| **Categorical**             | Gender, city, department, class |
| **Numerical + Numerical**   | `age` ↔ `salary`                |
| **Categorical + Numerical** | `department` ↔ `salary`         |
| **Categorical only**        | `gender`, `city`                |
| **Numerical Matrix**        | Correlation matrix              |

### ⭐ Quick Selection

```text
Trend                 → lineplot() / relplot(kind="line")
Relationship           → scatterplot() / relplot(kind="scatter")
Distribution           → histplot()/kdeplot() / displot()
Category comparison    → barplot() / catplot(kind="bar")
Outliers               → boxplot() / catplot(kind="box")
Category counting      → countplot() / catplot(kind="count")
Regression              → regplot() / lmplot()
Residuals               → residplot()
Matrix                  → heatmap() / clustermap()
Many features            → pairplot()
Two variables             → jointplot()
```

### Axes-level ↔ Figure-level Map

```text
scatterplot / lineplot   → relplot
histplot / kdeplot / ecdfplot → displot
barplot / boxplot / violinplot / stripplot / swarmplot / countplot / pointplot → catplot
regplot                    → lmplot
```

## Seaborn vs Matplotlib — 95% Daily-Use Parameters

| Purpose            | Seaborn                              | Matplotlib                           |
| ------------------ | ------------------------------------ | ------------------------------------ |
| **Data**           | `data=df`                            | `data=df` / arrays                   |
| **X-axis**         | `x="age"`                            | `x`                                  |
| **Y-axis**         | `y="salary"`                         | `y`                                  |
| **Group by color** | `hue="gender"`                       | `c=` / separate plots                |
| **Group by style** | `style="gender"`                     | Separate plots                       |
| **Point size**     | `size="salary"`                      | `s=50`                               |
| **Color**          | `color="blue"`                       | `color="blue"`                       |
| **Palette**        | `palette="viridis"`                  | `cmap="viridis"`                     |
| **Marker**         | `marker="o"`                         | `marker="o"`                         |
| **Transparency**   | `alpha=0.7`                          | `alpha=0.7`                          |
| **Line width**     | `linewidth=2`                        | `linewidth=2`                        |
| **Line style**     | `linestyle="--"`                     | `linestyle="--"`                     |
| **Bar width**      | `width=0.8`                          | `width=0.8`                          |
| **Histogram bins** | `bins=20`                            | `bins=20`                            |
| **Fill**           | `fill=True`                          | `fill=True` / `facecolor=`           |
| **Category order** | `order=[...]`                        | Manually control                     |
| **Hue order**      | `hue_order=[...]`                    | Manually control                     |
| **Error bars**     | `errorbar="sd"`                      | `yerr=` / `xerr=`                    |
| **Estimator**      | `estimator="mean"`                   | Calculate manually                   |
| **Legend**         | `legend=True`                        | `ax.legend()`                        |
| **Axes**           | `ax=ax`                              | `ax=ax`                              |
| **Figure size**    | `height=5`, `aspect=1.5`*            | `figsize=(10, 6)`                    |
| **Axis limits**    | `ax.set_xlim()`, `ax.set_ylim()`     | `ax.set_xlim()`, `ax.set_ylim()`     |
| **Labels**         | `ax.set_xlabel()`, `ax.set_ylabel()` | `ax.set_xlabel()`, `ax.set_ylabel()` |
| **Title**          | `ax.set_title()`                     | `ax.set_title()`                     |
| **Ticks**          | `ax.set_xticks()`, `ax.set_yticks()` | `ax.set_xticks()`, `ax.set_yticks()` |
| **Tick rotation**  | `ax.tick_params()`                   | `ax.tick_params()`                   |
| **Grid**           | `ax.grid()`                          | `ax.grid()`                          |
| **Log scale**      | `log_scale=True`*                    | `ax.set_xscale("log")`               |
| **Spines**         | `sns.despine()`                      | `ax.spines[]`                        |

* Mainly used with Seaborn **figure-level** functions such as `relplot()`, `displot()`, and `catplot()`.

### Quick rule

| If you want...         | Usually use              |
| ---------------------- | ------------------------ |
| Fast EDA               | **Seaborn**              |
| Statistical plots      | **Seaborn**              |
| Automatic grouping     | **Seaborn `hue=`**       |
| Detailed customization | **Matplotlib**           |
| Full figure control    | **Matplotlib**           |
| Both together          | **Seaborn + Matplotlib** |


## Importing Seaborn

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
````

````python
tips.sample(5)
````

## Relational Plots
- sns.lineplot()
- sns.scatterplot()
- sns.relplot()

````python
sns.lineplot(
    data=tips, 
    x="total_bill", 
    y="tip", 
    hue="sex", 
    )
````

````python
sns.scatterplot(
    data=tips,
    x="total_bill", 
    y="tip", hue="sex", 
    style="time", 
    size="size"
    )
````

````python
# facet plot
sns.relplot(
    data=tips, 
    x="total_bill", 
    y="tip", 
    kind="scatter", 
    col="smoker", 
    row="sex")
````
