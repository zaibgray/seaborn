## Seaborn
### Types of Functions
- Figure Level
- Axis Level

### Seaborn Cheat Sheet for ML & Research

*   **1. Relational Plots (Axis-Level)**
    *   `sns.lineplot()` -> Tracks loss or accuracy changes over time.
    *   `sns.scatterplot()` -> Shows how data points cluster together.

*   **2. Distribution Plots (Axis-Level)**
    *   `sns.histplot()` -> Shows data spread using bars.
    *   `sns.kdeplot()` -> Shows data spread using smooth curves.

*   **3. Categorical Plots (Axis-Level)**
    *   `sns.barplot()` -> Ranks model feature importance scores.
    *   `sns.boxplot()` -> Shows data ranges and catches outliers.
    *   `sns.countplot()` -> Counts items in each target class.

*   **4. Regression Plots (Axis-Level)**
    *   `sns.regplot()` -> Compares actual vs. predicted value trends.
    *   `sns.residplot()` -> Visualizes model prediction errors.

*   **5. Matrix Plots (Axis-Level)**
    *   `sns.heatmap()` -> Renders confusion matrices and correlation grids.

*   **6. Multi-plot Grids (Figure-Level)**
    *   `sns.pairplot()` -> Scans all feature relationships at once.
    *   `sns.jointplot()` -> Details the connection between two features.
